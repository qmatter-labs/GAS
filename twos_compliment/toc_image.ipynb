# TOC image

```python
import matplotlib.pyplot as plt
from gas.clock import create_binary_clock, add_hand, add_movement_arc
```

```python
plt.rcParams["font.family"] = "serif"
plt.rcParams["font.serif"] = ["Times New Roman"]
plt.rcParams['font.size'] = 16
```

```python
fig1, ax1, angles1, values1, x1, y1 = create_binary_clock(4, signed=True)

# Add two hands
hand1_idx = 2   
hand2_idx = 14   

add_hand(ax1, angles1[hand1_idx], length=0.8, color='green', width=4, label=None)
add_hand(ax1, angles1[hand2_idx], length=0.8, color='red', width=4, label=None)
# Anticlockwise from 1 to 4
add_movement_arc(ax1, angles1[hand1_idx], angles1[hand2_idx], 
                    radius=0.6, color='orange', linewidth=2.5,
                    clockwise=False, arrow_size=12, alpha=0.5)


hand1_idx = 7
hand2_idx = 3  
add_hand(ax1, angles1[hand1_idx], length=0.8, color='green', width=4, label=None)
add_hand(ax1, angles1[hand2_idx], length=0.8, color='grey', width=4, label=None)
# Anticlockwise from 1 to 4
add_movement_arc(ax1, angles1[hand1_idx], angles1[hand2_idx], 
                    radius=0.6, color='orange', linewidth=2.5,
                    clockwise=False, arrow_size=12, alpha=0.5)


plt.text(0, 0.65, "-y", color="orange", fontsize=25, fontweight='bold')
plt.text(0.62, -0.2, "-y", color="orange", fontsize=25, fontweight='bold')


plt.tight_layout()
# fig1.set_size_inches(3.25, 1.75) # JCTC specification
plt.savefig(f'toc_image_full_size.png', dpi=300)
plt.savefig(f'toc_image_full_size.tif', dpi=300)
plt.show()

```

```python
## resize image for JCTC

from PIL import Image

width_inches = 3.25
height_inches = 1.75
dpi = 300

target_width = round(width_inches * dpi)
target_height = round(height_inches * dpi)

img = Image.open("toc_image_full_size.tif")
img.thumbnail((target_width, target_height), Image.Resampling.LANCZOS)
img.save("toc_image.tif", dpi=(dpi, dpi))
```
