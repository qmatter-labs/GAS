# Draw binary clocks to explain Two's Compliment

```python
import matplotlib.pyplot as plt
from gas.clock import create_binary_clock, add_hand, add_movement_arc
```

```python
plt.rcParams["font.family"] = "serif"
plt.rcParams["font.serif"] = ["Times New Roman"]
plt.rcParams['font.size'] = 16
```

## Unsigned version

```python
# empty clock (UNSIGNED)
fig1, ax1, angles1, values1, x1, y1 = create_binary_clock(4, signed=False)
plt.tight_layout()
plt.savefig('empty_clock_unsigned.png', dpi=300)
```

```python
fig1, ax1, angles1, values1, x1, y1 = create_binary_clock(4, signed=False)
# Add two hands
hand1_idx = 2   
hand2_idx = 14

add_hand(ax1, angles1[hand1_idx], length=0.8, color='green', width=4, label=None)
add_hand(ax1, angles1[hand2_idx], length=0.8, color='red', width=4, label=None)

# Clockwise from 1 to 4
add_movement_arc(ax1, angles1[hand1_idx], angles1[hand2_idx], 
                    radius=0.4, color='purple', linewidth=3,
                    clockwise=True, arrow_size=12, alpha=0.5)

# Anticlockwise from 1 to 4
add_movement_arc(ax1, angles1[hand1_idx], angles1[hand2_idx], 
                    radius=0.6, color='orange', linewidth=2.5,
                    clockwise=False, arrow_size=12, alpha=0.5)


plt.tight_layout()
plt.savefig('clock_unsigned_2_14.png', dpi=300)
plt.show()
```
