```python
from gas.dicke import dicke_state
from qiskit import transpile
from qiskit_aer import StatevectorSimulator

import numpy as np

from symmer.operators import QuantumState
```

```python
# define Dicke state
n_qubits = 5
hamming_weight = 2
assert hamming_weight<=n_qubits, "hamming weight larger than number of qubits"
```

```python
# build QC
dicke_qc= dicke_state(n_qubits, hamming_weight)
full_dicke = transpile(dicke_qc, basis_gates=['id', 'ry', 'rx', 'cx'], optimization_level=1, seed_transpiler=1) 
```

```python
# Statevector Simulator 
simulator_aerlegacy = StatevectorSimulator()
result_aerlegacy = simulator_aerlegacy.run(full_dicke.decompose().decompose())

statevec_dicke = result_aerlegacy.result().get_statevector().data.reshape([-1, 1])
statevec_dicke[np.abs(statevec_dicke)<1e-15] = 0 # zero out vanishingly small terms!
```

```python
psi_dicke = QuantumState.from_array(statevec_dicke)
psi_dicke
```
